## Ejercicio 3: Inspección visual de un producto
Problema: en una línea de producción, una cámara captura una pequeña imagen de 5×5 píxeles
de cada producto, representada como una matriz de NumPy donde cada valor es 0 (píxel normal)
o 1 (píxel defectuoso). El agente debe decidir qué hacer con el producto según cuántos píxeles
defectuosos detecta.
### Ficha PEAS

| Elemento | Descripción |
|---|---|
| **Percepción (S)** | Matriz 5×5 con valores 0 (normal) o 1 (defectuoso) por píxel |
| **Acciones (A)** | Aprobar, enviar a revisión manual, rechazar |
| **Entorno (E)** | Línea de producción con productos de distinta calidad pasando frente a la cámara |
| **Objetivo** | Filtrar productos defectuosos antes de que lleguen al cliente, sin descartar productos buenos |
| **Medida de desempeño (P)** | Productos defectuosos no detectados (fuga), productos buenos rechazados de más, tiempo de inspección |
