## Ejercicio 2: Escalamiento de tickets de soporte técnico
Problema: un sistema de mesa de ayuda decide a qué nivel de soporte enviar un ticket,
considerando:
1. tiempo_espera_minutos: un número entero
2. nivel_urgencia: un texto, puede ser "baja", "media" o "alta"
3. cliente_premium: True o False
### Ficha PEAS

| Elemento | Descripción |
|---|---|
| **Percepción (S)** | Tiempo de espera del ticket, nivel de urgencia declarado, si el cliente es premium |
| **Acciones (A)** | Mantener en nivel actual, escalar a nivel 2, escalar a nivel 3 |
| **Entorno (E)** | Mesa de ayuda de soporte técnico, con tickets de distinta prioridad y tipo de cliente |
| **Objetivo** | Atender primero los casos más urgentes o críticos, evitando esperas excesivas |
| **Medida de desempeño (P)** | Tiempo promedio de resolución, satisfacción del cliente, tickets urgentes no atendidos a tiempo |

### Justificación de las reglas de decisión
La urgencia alta se escala de inmediato porque puede representar un problema crítico para el cliente. En la urgencia media, la espera aumenta la prioridad: si supera los 30 minutos, se considera como una urgencia alta; si el cliente es premium, se prioriza aunque la espera sea moderada, según el acuerdo establecido. En la urgencia baja, solo se escala cuando un cliente premium ha esperado demasiado, evitando que casos simples saturen los niveles superiores de soporte.

| Condición | Acción |
|---|---|
| Urgencia alta | Escalar a nivel 3 |
| Urgencia media, espera > 30 min | Escalar a nivel 3 |
| Urgencia media, espera > 15 min o cliente premium | Escalar a nivel 2 |
| Urgencia media, espera corta y no premium | Mantener en nivel actual |
| Urgencia baja, cliente premium con espera > 30 min | Escalar a nivel 2 |
| Urgencia baja, resto de casos | Mantener en nivel actual |

### Código

```python
def agente_soporte(tiempo_espera_minutos, nivel_urgencia, cliente_premium):
    if nivel_urgencia == "alta":
        return "escalar a nivel 3", "urgencia alta reportada por el cliente"
    elif nivel_urgencia == "media":
        if tiempo_espera_minutos > 30:
            return "escalar a nivel 3", "urgencia media con tiempo de espera mayor a 30 minutos"
        elif tiempo_espera_minutos > 15 or cliente_premium:
            return "escalar a nivel 2", "urgencia media con espera considerable o cliente premium"
        else:
            return "mantener en nivel actual", "urgencia media con espera corta y cliente no premium"
    else:  # urgencia baja
        if cliente_premium and tiempo_espera_minutos > 30:
            return "escalar a nivel 2", "urgencia baja pero cliente premium con espera prolongada"
        else:
            return "mantener en nivel actual", "urgencia baja sin condiciones que ameriten escalamiento"
```
